```python
import sys
print(sys.version)
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import scipy.optimize
import numpy as np

print("Imports Finished!")
```

```python
data = pd.read_excel('New Microsoft Excel Worksheet.xlsx')
data.head()
```

```python
def ll5p(x, b, c, d, e, f):
    '''
    given data, x, and parameters b-f
    return fitted value for a 5 parameter log-logistic model    
    '''
    return (c + (d-c)/(1 + np.exp(b*(np.log(x) - np.log(e))))**f)
    
```

```python
param, pcov = scipy.optimize.curve_fit(ll5p, data.t, data.phi)
param
```

```python
plt.plot(data.t, data.phi, 'bo', label = 'tree 47-40')
plt.plot(data.t, ll5p(data.t, *param), 'r-')
plt.show()
```

Hey this is a model of the tree angles

$$f(x) = c + \frac{d-c}{(1-exp(b(log(x)-log(e))))^f)}$$
